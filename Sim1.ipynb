```python
# ================================================================
# EXPERIMENT 1
# Observed-record fit versus hidden missed-failure risk
#
# Four estimators
#   phase_likelihood   Select duration shapes and estimate means
#   known_shapes       Fix the true shapes and estimate means
#   exponential        Restrict both durations to exponential laws
#   age_reset          Incorrectly restart age after every inspection
#
# Additional analyses
#   Population-likelihood fitting without sampling variability
#   Within-shape optimization diagnostics
#   Exploratory near-optimal risk ranges
#   Paired comparisons and Monte Carlo standard errors
#
# Independent Colab cell. No Google Drive connection.
# Local checkpoints retain completed tasks.
# An interrupted task is repeated with the same random seed.
# ================================================================

import os
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")

import json
import gzip
import time
import hashlib
import shutil
import platform
import itertools
import uuid
from pathlib import Path
from functools import lru_cache

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy
from scipy.linalg import expm, expm_frechet, block_diag
from scipy.optimize import minimize


# ================================================================
# Configuration
# ================================================================

PROFILE = "publication"       # "diagnostic" or "publication"
SEED = 20261116
CHECKPOINT_SECONDS = 20 * 60
DOWNLOAD_ON_FINISH = True
RUN_VALIDATION = True

CODE_VERSION = "experiment1_diagnostic_framework_v2"

PROFILES = {
    "diagnostic": {
        "reps": 3,
        "sample_sizes": [200, 800],
        "starts_per_shape": 6,
        "coarse_grid_side": 5,
        "maxiter": 250,
        "refine_maxiter": 500,
        "validation_systems": 20000,
    },
    "publication": {
        "reps": 200,
        "sample_sizes": [200, 800, 3200],
        "starts_per_shape": 10,
        "coarse_grid_side": 5,
        "maxiter": 400,
        "refine_maxiter": 800,
        "validation_systems": 50000,
    },
}

if PROFILE not in PROFILES:
    raise ValueError("PROFILE must be diagnostic or publication.")

CFG = PROFILES[PROFILE].copy()

TIMES = np.array([0.0, 1.5, 3.0, 4.5, 6.0])
SHAPES = (0, 1, 3)
SHAPE_PAIRS = tuple(itertools.product(SHAPES, repeat=2))
MEAN_BOUNDS = (0.12, 30.0)
LOG_BOUNDS = np.log(np.array(MEAN_BOUNDS))
DELTA = float(TIMES[-1] - TIMES[-2])

# Shape 0 is a fixed two-component hyperexponential family.
# Shape 1 is exponential.
# Shape 3 is Erlang with three phases.
#
# All systems start newly in operation.
# Repairs restore a fresh operating duration.
# Inspections reveal only the binary state and do not alter the system.

SCENARIOS = [
    {
        "scenario": f"k0_{k0}_repair_{mean1}",
        "k0": k0,
        "k1": 3,
        "mean0": 4.0,
        "mean1": mean1,
    }
    for k0 in SHAPES
    for mean1 in (0.3, 1.2)
]

METHODS = (
    "phase_likelihood",
    "known_shapes",
    "exponential",
    "age_reset",
)

METHOD_LABELS = {
    "phase_likelihood": "Estimated shapes",
    "known_shapes": "Known shapes",
    "exponential": "Exponential restriction",
    "age_reset": "Age reset",
}

COLORS = {
    "phase_likelihood": "#0072B2",
    "known_shapes": "#CC79A7",
    "exponential": "#D55E00",
    "age_reset": "#009E73",
}

# These are descriptive likelihood-distance thresholds.
# They are NOT calibrated confidence thresholds.
SAMPLE_TWO_LOG_LIKELIHOOD_GAPS = (2.0, 6.0, 10.0)

# These define approximate population observational similarity.
# They do NOT establish exact observational equivalence.
POPULATION_KL_THRESHOLDS = (1e-8, 1e-5, 1e-3)

BOUNDARY_TOLERANCE = 1e-5
PROJECTED_GRADIENT_TOLERANCE = 1e-5

# Numerical agreement tolerance for per-system objective values.
OBJECTIVE_AGREEMENT_TOLERANCE = 1e-7

# Underestimation by at least this absolute risk difference.
UNDERREPORTING_THRESHOLD = 0.10

BASE = (
    Path("/content") if Path("/content").exists() else Path.cwd()
) / "SemiMarkov_Experiments"

# A separate directory preserves earlier Experiment 1 results.
ROOT = BASE / "experiment_1_strengthened"

CP = None


# ================================================================
# Local checkpoints and reproducibility
# ================================================================

def json_default(value):
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(f"Cannot serialize {type(value)}")


def reproducible_rng(key):
    digest = hashlib.sha256(f"{SEED}|{key}".encode()).digest()
    return np.random.default_rng(int.from_bytes(digest[:8], "little"))


class LocalCheckpoint:
    def __init__(self, root, metadata):
        self.root = Path(root)
        self.metadata = metadata
        self.fingerprint = hashlib.sha256(
            json.dumps(metadata, sort_keys=True).encode()
        ).hexdigest()

        choice = input("Enter 1 to START NEW or 2 to RESUME  ").strip()
        while choice not in {"1", "2"}:
            choice = input("Please enter 1 or 2  ").strip()

        self.root.parent.mkdir(parents=True, exist_ok=True)

        if choice == "1" and self.root.exists():
            archived = self.root.with_name(
                self.root.name
                + "_archive_"
                + time.strftime("%Y%m%d_%H%M%S")
                + "_"
                + uuid.uuid4().hex[:6]
            )
            shutil.move(str(self.root), str(archived))
            print("Previous local run preserved in", archived)

        self.root.mkdir(parents=True, exist_ok=True)
        self.path = self.root / "checkpoint.json.gz"

        if choice == "2":
            if self.path.exists():
                raw = self.path.read_bytes()
            else:
                print("Upload a checkpoint produced by this code version.")
                try:
                    from google.colab import files
                except ImportError as exc:
                    raise FileNotFoundError(
                        f"Place checkpoint.json.gz in {self.root}"
                    ) from exc

                uploaded = files.upload()
                candidates = [
                    data
                    for name, data in uploaded.items()
                    if name.endswith(".json.gz")
                ]
                if len(candidates) != 1:
                    raise ValueError("Upload exactly one .json.gz checkpoint.")
                raw = candidates[0]

            state = json.loads(gzip.decompress(raw).decode("utf-8"))
            if state.get("fingerprint") != self.fingerprint:
                raise ValueError(
                    "The checkpoint belongs to a different configuration "
                    "or code version. Restore matching settings or start new."
                )
            if not isinstance(state.get("tasks"), dict):
                raise ValueError("Invalid checkpoint task table.")
            self.state = state
            print("Resuming", len(self.state["tasks"]), "completed tasks.")
        else:
            self.state = {
                "fingerprint": self.fingerprint,
                "metadata": metadata,
                "tasks": {},
            }

        self.last_save = time.monotonic()
        self.save()

    def has(self, key):
        return key in self.state["tasks"]

    def commit(self, key, tables):
        # Only fully completed tasks enter the resumable state.
        self.state["tasks"][key] = tables
        self.tick()

    def save(self):
        self.state["saved_at_utc"] = time.strftime(
            "%Y-%m-%dT%H:%M:%SZ", time.gmtime()
        )
        raw = json.dumps(
            self.state,
            default=json_default,
            allow_nan=True,
        ).encode("utf-8")

        temporary = self.root / "checkpoint.tmp.gz"
        with gzip.open(temporary, "wb") as stream:
            stream.write(raw)
        os.replace(temporary, self.path)

        self.last_save = time.monotonic()
        print(
            "Checkpoint saved",
            len(self.state["tasks"]),
            "completed tasks",
            flush=True,
        )

    def tick(self):
        if time.monotonic() - self.last_save >= CHECKPOINT_SECONDS:
            self.save()

    def frame(self, table):
        rows = []
        for tables in self.state["tasks"].values():
            rows.extend(tables.get(table, []))
        return pd.DataFrame(rows)

    def export(self):
        names = {
            name
            for tables in self.state["tasks"].values()
            for name in tables
        }
        for name in sorted(names):
            self.frame(name).to_csv(self.root / f"{name}.csv", index=False)

        (self.root / "configuration.json").write_text(
            json.dumps(self.metadata, indent=2),
            encoding="utf-8",
        )


def heartbeat():
    if CP is not None:
        CP.tick()


def download_checkpoint():
    """Call in another cell after stopping execution to save progress."""
    if CP is None:
        raise RuntimeError("Checkpoint manager has not been initialized.")
    CP.save()
    try:
        from google.colab import files
        files.download(str(CP.path))
    except ImportError:
        print("Checkpoint", CP.path)


def download_results():
    """Download completed results and the checkpoint in one ZIP."""
    if CP is not None:
        CP.save()
        CP.export()
    archive = shutil.make_archive(str(ROOT) + "_results", "zip", ROOT)
    try:
        from google.colab import files
        files.download(archive)
    except ImportError:
        print("Results archive", archive)


# ================================================================
# Physical model
# ================================================================

def duration_block(shape, mean):
    if mean <= 0:
        raise ValueError("Duration means must be positive.")

    if shape == 0:
        rates = np.array([0.6, 3.0]) * (0.5 / 0.6 + 0.5 / 3.0) / mean
        return np.array([0.5, 0.5]), -np.diag(rates)

    alpha = np.zeros(shape)
    alpha[0] = 1.0
    T = -np.eye(shape) * shape / mean
    if shape > 1:
        T[np.arange(shape - 1), np.arange(1, shape)] = shape / mean
    return alpha, T


@lru_cache(maxsize=2048)
def physical_model(k0, k1, mean0, mean1):
    a, T = duration_block(k0, mean0)
    b, U = duration_block(k1, mean1)

    Q = block_diag(T, U)
    d = len(a)

    Q[:d, d:] = np.outer(-T.sum(axis=1), b)
    Q[d:, :d] = np.outer(-U.sum(axis=1), a)

    alpha = np.r_[a, np.zeros(len(b))]
    states = np.r_[np.zeros(d, dtype=int), np.ones(len(b), dtype=int)]

    fresh0 = np.r_[a, np.zeros(len(b))]
    fresh1 = np.r_[np.zeros(len(a)), b]

    # Derivatives with respect to log(mean0) and log(mean1).
    dQ0 = np.zeros_like(Q)
    dQ1 = np.zeros_like(Q)
    dQ0[:d, :] = -Q[:d, :]
    dQ1[d:, :] = -Q[d:, :]

    return Q, alpha, states, T, a, fresh0, fresh1, dQ0, dQ1


def all_records(number):
    return np.array(
        list(itertools.product((0, 1), repeat=number)),
        dtype=int,
    )


def encode(readings):
    return readings @ (2 ** np.arange(readings.shape[1] - 1, -1, -1))


# Earlier records that end in operation immediately before the last visit.
EARLIER_RECORDS = all_records(len(TIMES) - 2)
TARGET_HISTORIES = [
    tuple(int(y) for y in history)
    for history in EARLIER_RECORDS
    if history[-1] == 0
]
HISTORY_NAMES = [
    "0" + "".join(map(str, history)) + "0"
    for history in TARGET_HISTORIES
]


# ================================================================
# Record probabilities and analytical likelihood derivatives
# ================================================================

def observation_law(parameters, reset=False, derivatives=False):
    heartbeat()

    (
        Q, alpha, states, T, a,
        fresh0, fresh1, dQ0, dQ1,
    ) = physical_model(*tuple(parameters))

    rows = alpha[None, :]
    if derivatives:
        derivative_rows = np.zeros((2, 1, len(alpha)))

    transitions = {}

    for dt in np.diff(TIMES):
        dt = float(dt)

        if dt not in transitions:
            P = expm(Q * dt)
            if derivatives:
                dP = np.stack([
                    expm_frechet(Q * dt, dQ0 * dt, compute_expm=False),
                    expm_frechet(Q * dt, dQ1 * dt, compute_expm=False),
                ])
                transitions[dt] = (P, dP)
            else:
                transitions[dt] = P

        if derivatives:
            P, dP = transitions[dt]
            prediction = rows @ P
            prediction_derivatives = np.stack([
                derivative_rows[j] @ P + rows @ dP[j]
                for j in range(2)
            ])
        else:
            P = transitions[dt]
            prediction = rows @ P

        updated = np.empty((2 * len(rows), len(alpha)))
        if derivatives:
            updated_derivatives = np.empty(
                (2, 2 * len(rows), len(alpha))
            )

        for state, fresh in ((0, fresh0), (1, fresh1)):
            mask = states == state
            restricted = prediction * mask

            if derivatives:
                restricted_derivatives = prediction_derivatives * mask

            if reset:
                restricted = restricted.sum(axis=1)[:, None] * fresh
                if derivatives:
                    restricted_derivatives = (
                        restricted_derivatives.sum(axis=2)[:, :, None]
                        * fresh[None, None, :]
                    )

            updated[state::2] = restricted

            if derivatives:
                updated_derivatives[:, state::2] = restricted_derivatives

        rows = updated
        if derivatives:
            derivative_rows = updated_derivatives

    probabilities = rows.sum(axis=1)

    if probabilities.min() < -1e-9:
        raise FloatingPointError("Negative observation probability.")
    if abs(probabilities.sum() - 1.0) > 1e-8:
        raise FloatingPointError("Observation probabilities do not sum to one.")

    probabilities = np.maximum(probabilities, 0.0)

    if derivatives:
        gradient = derivative_rows.sum(axis=2).T
        return probabilities, gradient

    return probabilities


def objective_value(parameters, frequencies, reset=False):
    probabilities = observation_law(parameters, reset=reset)
    return -float(
        frequencies @ np.log(np.maximum(probabilities, 1e-300))
    )


def objective_with_gradient(log_means, shape_pair, frequencies, reset):
    parameters = (
        shape_pair[0],
        shape_pair[1],
        *np.exp(log_means),
    )
    probabilities, gradient = observation_law(
        parameters, reset=reset, derivatives=True
    )
    safe = np.maximum(probabilities, 1e-300)
    value = -float(frequencies @ np.log(safe))
    derivative = -np.sum(
        (frequencies / safe)[:, None] * gradient,
        axis=0,
    )
    return value, derivative


def record_kl(truth, fitted):
    value = float(
        truth @ np.log(
            np.maximum(truth, 1e-300) / np.maximum(fitted, 1e-300)
        )
    )
    if value < -1e-8:
        raise FloatingPointError("Materially negative KL divergence.")
    return max(0.0, value)


# ================================================================
# Hidden final-interval risk
# ================================================================

def conditional_risks(parameters, reset=False):
    (
        Q, alpha, states, T, a,
        fresh0, fresh1, _, _,
    ) = physical_model(*tuple(parameters))

    transition = expm(Q * DELTA)
    uninterrupted_transition = expm(T * DELTA)

    values = []
    masses = []

    for history in TARGET_HISTORIES:
        distribution = alpha.copy()

        for reading in history:
            distribution = (
                distribution @ transition
            ) * (states == reading)

            if reset:
                fresh = fresh0 if reading == 0 else fresh1
                distribution = distribution.sum() * fresh

        history_mass = float(distribution.sum())
        if history_mass <= 0:
            raise FloatingPointError("A target history has zero probability.")

        conditional = distribution / history_mass

        p = float(
            (conditional @ transition)[states == 0].sum()
        )
        q = float(
            conditional[states == 0]
            @ uninterrupted_transition
            @ np.ones(len(a))
        )

        if p <= 0:
            raise FloatingPointError("Zero operating-endpoint probability.")

        raw_risk = 1.0 - q / p
        if raw_risk < -1e-7 or raw_risk > 1.0 + 1e-7:
            raise FloatingPointError("Risk outside probability range.")

        values.append(float(np.clip(raw_risk, 0.0, 1.0)))
        masses.append(history_mass * p)

    return np.array(values), np.array(masses)


def risk_metrics(estimates, truth, weights):
    error = estimates - truth
    mse = float(weights @ (error ** 2))

    return {
        "risk_MSE": mse,
        "risk_RMSE": float(np.sqrt(mse)),
        "risk_bias": float(weights @ error),
        "risk_MAE": float(weights @ np.abs(error)),
        "underreporting_mass": float(
            weights @ (error <= -UNDERREPORTING_THRESHOLD)
        ),
        "max_history_error": float(np.max(np.abs(error))),
    }


# ================================================================
# Independent continuous-time simulation
# ================================================================

def draw_duration(shape, mean, rng):
    if shape == 0:
        rates = np.array([0.6, 3.0]) * (0.5 / 0.6 + 0.5 / 3.0) / mean
        return rng.exponential(1.0 / rng.choice(rates))
    return rng.gamma(shape, mean / shape)


def simulate(parameters, n, rng):
    k0, k1, mean0, mean1 = parameters

    readings = np.zeros((n, len(TIMES) - 1), dtype=int)
    final_interval_failure = np.zeros(n, dtype=bool)

    for system in range(n):
        if system % 100 == 0:
            heartbeat()

        current_time = 0.0
        state = 0
        transitions = []

        while current_time < TIMES[-1]:
            current_time += draw_duration(
                k0 if state == 0 else k1,
                mean0 if state == 0 else mean1,
                rng,
            )

            if current_time > TIMES[-1]:
                break

            state = 1 - state
            transitions.append(current_time)

            if state == 1 and current_time > TIMES[-2]:
                final_interval_failure[system] = True

        readings[system] = np.searchsorted(
            transitions, TIMES[1:], side="right"
        ) % 2

    return readings, final_interval_failure


# ================================================================
# Optimization
# ================================================================

def at_boundary(log_means):
    return bool(np.any(
        (log_means - LOG_BOUNDS[0] <= BOUNDARY_TOLERANCE)
        | (LOG_BOUNDS[1] - log_means <= BOUNDARY_TOLERANCE)
    ))


def projected_gradient_norm(log_means, gradient):
    projected = np.asarray(gradient, dtype=float).copy()

    at_lower = log_means - LOG_BOUNDS[0] <= BOUNDARY_TOLERANCE
    at_upper = LOG_BOUNDS[1] - log_means <= BOUNDARY_TOLERANCE

    projected[at_lower & (projected > 0)] = 0.0
    projected[at_upper & (projected < 0)] = 0.0

    return float(np.max(np.abs(projected)))


def candidate_parameters(shape_pair, log_means):
    return (
        int(shape_pair[0]),
        int(shape_pair[1]),
        float(np.exp(log_means[0])),
        float(np.exp(log_means[1])),
    )


def fit_shape_pair(frequencies, shape_pair, reset, key):
    # Coarse evaluations help choose starts without using the true parameters.
    grid_axis = np.linspace(
        LOG_BOUNDS[0], LOG_BOUNDS[1], CFG["coarse_grid_side"]
    )
    grid_points = [
        np.array(z, dtype=float)
        for z in itertools.product(grid_axis, repeat=2)
    ]

    coarse = []
    for z in grid_points:
        value = objective_value(
            candidate_parameters(shape_pair, z),
            frequencies,
            reset,
        )
        coarse.append({
            "z": z,
            "objective": value,
            "source": "coarse_grid",
            "success": False,
            "projected_gradient": np.nan,
            "nit": 0,
            "message": "Unoptimized coarse-grid candidate",
        })

    ordered = sorted(coarse, key=lambda item: item["objective"])

    starts = []
    number_informed = max(2, CFG["starts_per_shape"] // 2)

    for candidate in ordered:
        z = candidate["z"]
        if all(np.linalg.norm(z - previous) > 0.5 for previous in starts):
            starts.append(z.copy())
        if len(starts) >= number_informed:
            break

    rng = reproducible_rng(f"starts|{key}|{shape_pair}|{reset}")
    while len(starts) < CFG["starts_per_shape"]:
        starts.append(rng.uniform(LOG_BOUNDS[0], LOG_BOUNDS[1], size=2))

    runs = []

    def optimize(z, source, maxiter, ftol, gtol):
        result = minimize(
            objective_with_gradient,
            z,
            args=(shape_pair, frequencies, reset),
            jac=True,
            method="L-BFGS-B",
            bounds=[tuple(LOG_BOUNDS)] * 2,
            options={
                "maxiter": maxiter,
                "ftol": ftol,
                "gtol": gtol,
                "maxls": 50,
            },
        )

        if not np.isfinite(result.fun):
            raise FloatingPointError("Nonfinite optimization objective.")

        value, gradient = objective_with_gradient(
            result.x, shape_pair, frequencies, reset
        )

        return {
            "z": np.asarray(result.x, dtype=float),
            "objective": float(value),
            "source": source,
            "success": bool(result.success),
            "projected_gradient": projected_gradient_norm(result.x, gradient),
            "nit": int(result.nit),
            "message": str(result.message),
        }

    for index, start in enumerate(starts):
        runs.append(optimize(
            start,
            f"multistart_{index}",
            CFG["maxiter"],
            1e-12,
            1e-8,
        ))

    # Refine two distinct promising solutions.
    refinement_starts = []
    for candidate in sorted(runs, key=lambda item: item["objective"]):
        if all(
            np.linalg.norm(candidate["z"] - previous) > 1e-3
            for previous in refinement_starts
        ):
            refinement_starts.append(candidate["z"].copy())
        if len(refinement_starts) == 2:
            break

    for index, start in enumerate(refinement_starts):
        runs.append(optimize(
            start,
            f"refinement_{index}",
            CFG["refine_maxiter"],
            1e-14,
            1e-10,
        ))

    # Retaining coarse candidates prevents numerical optimization from
    # replacing an already better evaluated point with a worse point.
    pool = coarse + runs
    best = min(pool, key=lambda item: item["objective"])

    run_values = np.array([run["objective"] for run in runs])
    within_shape_spread = float(run_values.max() - run_values.min())
    agreement_fraction = float(np.mean(
        run_values - best["objective"] <= OBJECTIVE_AGREEMENT_TOLERANCE
    ))

    for candidate in pool:
        candidate["parameters"] = candidate_parameters(
            shape_pair, candidate["z"]
        )

    return {
        "shape_pair": shape_pair,
        "best": best,
        "pool": pool,
        "runs": runs,
        "within_shape_spread": within_shape_spread,
        "agreement_fraction": agreement_fraction,
    }


def fit_family(frequencies, reset, key):
    fits = {}
    for shape_pair in SHAPE_PAIRS:
        heartbeat()
        fits[shape_pair] = fit_shape_pair(
            frequencies, shape_pair, reset, key
        )
    return fits


def select_methods(correct_fits, reset_fits, true_shapes):
    selected_pair = min(
        correct_fits,
        key=lambda pair: correct_fits[pair]["best"]["objective"],
    )
    reset_pair = min(
        reset_fits,
        key=lambda pair: reset_fits[pair]["best"]["objective"],
    )

    return {
        "phase_likelihood": (correct_fits[selected_pair], False),
        "known_shapes": (correct_fits[true_shapes], False),
        "exponential": (correct_fits[(1, 1)], False),
        "age_reset": (reset_fits[reset_pair], True),
    }


def optimizer_rows(fits, context, reset):
    rows = []

    for shape_pair, fit in fits.items():
        best_objective = fit["best"]["objective"]

        for index, run in enumerate(fit["runs"]):
            parameters = run["parameters"]
            rows.append({
                **context,
                "reset": reset,
                "shape0": shape_pair[0],
                "shape1": shape_pair[1],
                "run_index": index,
                "source": run["source"],
                "mean0": parameters[2],
                "mean1": parameters[3],
                "objective_per_system": run["objective"],
                "gap_from_shape_best": run["objective"] - best_objective,
                "success": run["success"],
                "projected_gradient": run["projected_gradient"],
                "boundary": at_boundary(run["z"]),
                "iterations": run["nit"],
                "message": run["message"],
                "within_shape_spread": fit["within_shape_spread"],
                "agreement_fraction": fit["agreement_fraction"],
            })

    return rows


# ================================================================
# Exploratory risk ranges
# ================================================================

def explored_risk_ranges(
    fits, truth_risks, weights, context,
    population=False, true_parameters=None, truth_probabilities=None,
):
    # We only use correctly age-aware candidates here.
    # Endpoints are taken over EVALUATED candidates.
    # This is not optimization over the entire parameter space.

    candidates = []
    for fit in fits.values():
        candidates.extend(fit["pool"])

    if population:
        thresholds = POPULATION_KL_THRESHOLDS

        def distance(candidate):
            predicted = observation_law(candidate["parameters"])
            return record_kl(truth_probabilities, predicted)

        distances = np.array([distance(candidate) for candidate in candidates])

        # The known generating model is a population reference only.
        # It is never inserted into a sample estimator or its starting values.
        candidates.append({
            "parameters": true_parameters,
            "source": "known_population_reference",
        })
        distances = np.r_[distances, 0.0]
        threshold_name = "population_KL"
    else:
        thresholds = SAMPLE_TWO_LOG_LIKELIHOOD_GAPS
        best = min(candidate["objective"] for candidate in candidates)
        n = context["n"]
        distances = np.array([
            2 * n * (candidate["objective"] - best)
            for candidate in candidates
        ])
        threshold_name = "two_log_likelihood_gap"

    history_rows = []
    aggregate_rows = []

    # Only compute risks for candidates needed by at least one threshold.
    largest_threshold = max(thresholds)
    candidate_risks = {}
    seen_parameters = {}

    for index, (candidate, distance) in enumerate(zip(candidates, distances)):
        if distance <= largest_threshold + 1e-10:
            parameter_key = tuple(candidate["parameters"])
            if parameter_key not in seen_parameters:
                seen_parameters[parameter_key] = conditional_risks(
                    parameter_key
                )[0]
            candidate_risks[index] = seen_parameters[parameter_key]

    for threshold in thresholds:
        selected_indices = [
            index for index, distance in enumerate(distances)
            if distance <= threshold + 1e-10
        ]
        if not selected_indices:
            continue

        matrix = np.vstack([
            candidate_risks[index] for index in selected_indices
        ])
        lower = matrix.min(axis=0)
        upper = matrix.max(axis=0)

        for j, history in enumerate(HISTORY_NAMES):
            history_rows.append({
                **context,
                "threshold_type": threshold_name,
                "threshold": threshold,
                "history": history,
                "true_risk": truth_risks[j],
                "explored_lower": lower[j],
                "explored_upper": upper[j],
                "explored_width": upper[j] - lower[j],
                "truth_inside_explored_range": bool(
                    lower[j] - 1e-10 <= truth_risks[j] <= upper[j] + 1e-10
                ),
                "candidate_count": len(selected_indices),
            })

        aggregate_rows.append({
            **context,
            "threshold_type": threshold_name,
            "threshold": threshold,
            "weighted_explored_width": float(weights @ (upper - lower)),
            "maximum_explored_width": float(np.max(upper - lower)),
            "candidate_count": len(selected_indices),
        })

    return history_rows, aggregate_rows


# ================================================================
# Validation
# ================================================================

def run_validation():
    key = "validation"
    if CP.has(key):
        print("Using previously completed validation checks.")
        return

    rows = []

    # 1. Closed-form exponential observation law.
    parameters = (1, 1, 4.0, 0.7)
    probabilities = observation_law(parameters)

    lam, mu = 1 / 4.0, 1 / 0.7
    failed_at_first = lam / (lam + mu) * (
        1 - np.exp(-(lam + mu) * DELTA)
    )
    full_records = all_records(len(TIMES) - 1)
    computed_failed_at_first = probabilities[
        full_records[:, 0] == 1
    ].sum()

    error = abs(computed_failed_at_first - failed_at_first)
    assert error < 1e-11
    rows.append({"check": "exponential_first_reading", "error": error})

    # 2. Under exponential durations, age resetting has no effect.
    error = float(np.max(abs(
        observation_law(parameters, reset=False)
        - observation_law(parameters, reset=True)
    )))
    assert error < 1e-11
    rows.append({"check": "exponential_reset_equivalence", "error": error})

    # 3. Analytical gradients against central finite differences.
    test_parameters = (3, 0, 4.0, 0.7)
    frequencies = observation_law((0, 3, 3.5, 1.0))
    log_means = np.log(test_parameters[2:])
    step = 1e-5

    for reset in (False, True):
        _, analytic = objective_with_gradient(
            log_means,
            test_parameters[:2],
            frequencies,
            reset,
        )
        numerical = []

        for j in range(2):
            direction = np.zeros(2)
            direction[j] = step

            upper = objective_value(
                candidate_parameters(test_parameters[:2], log_means + direction),
                frequencies,
                reset,
            )
            lower = objective_value(
                candidate_parameters(test_parameters[:2], log_means - direction),
                frequencies,
                reset,
            )
            numerical.append((upper - lower) / (2 * step))

        error = float(np.max(abs(analytic - np.array(numerical))))
        assert error < 2e-5
        rows.append({
            "check": f"likelihood_gradient_reset_{reset}",
            "error": error,
        })

    # 4. Continuous-time simulation versus exact probabilities.
    n = CFG["validation_systems"]
    observed, hidden = simulate(
        test_parameters, n, reproducible_rng("validation_paths")
    )
    truth = observation_law(test_parameters)
    empirical = np.bincount(
        encode(observed), minlength=len(truth)
    ) / n

    standard_errors = np.sqrt(truth * (1 - truth) / n)
    tolerances = 6 * standard_errors + 6 / n
    assert np.all(abs(empirical - truth) <= tolerances)

    rows.append({
        "check": "record_probabilities_against_paths",
        "error": float(np.max(abs(empirical - truth))),
    })

    # 5. Direct final-interval failure indicators versus risk formula.
    true_risks, _ = conditional_risks(test_parameters)

    for j, history in enumerate(TARGET_HISTORIES):
        full_history = np.array(history + (0,), dtype=int)
        selected = np.all(observed == full_history, axis=1)
        count = int(selected.sum())

        if count >= 100:
            empirical_risk = float(hidden[selected].mean())
            standard_error = np.sqrt(
                true_risks[j] * (1 - true_risks[j]) / count
            )
            assert abs(empirical_risk - true_risks[j]) <= (
                6 * standard_error + 6 / count
            )
            rows.append({
                "check": f"conditional_risk_{HISTORY_NAMES[j]}",
                "error": abs(empirical_risk - true_risks[j]),
                "selected_systems": count,
            })

    CP.commit(key, {"validation": rows})
    CP.save()
    print("All validation checks passed.")


# ================================================================
# Population experiment
# ================================================================

def run_population_scenario(scenario):
    name = scenario["scenario"]
    key = f"population|{name}"

    if CP.has(key):
        return

    true_parameters = (
        scenario["k0"], scenario["k1"],
        scenario["mean0"], scenario["mean1"],
    )
    true_shapes = true_parameters[:2]
    truth_probabilities = observation_law(true_parameters)
    true_risks, masses = conditional_risks(true_parameters)
    weights = masses / masses.sum()

    print("Population analysis", name, flush=True)
    correct = fit_family(truth_probabilities, False, key)
    reset = fit_family(truth_probabilities, True, key)
    selected = select_methods(correct, reset, true_shapes)

    context = {
        "scenario": name,
        "k0": scenario["k0"],
        "repair_mean": scenario["mean1"],
        "stage": "population",
        "n": 0,
        "rep": -1,
    }

    accuracy_rows = []
    history_rows = []

    for method, (fit, use_reset) in selected.items():
        best = fit["best"]
        parameters = best["parameters"]
        predicted = observation_law(parameters, reset=use_reset)
        estimated_risks, _ = conditional_risks(
            parameters, reset=use_reset
        )
        divergence = record_kl(truth_probabilities, predicted)

        accuracy_rows.append({
            **context,
            "method": method,
            "record_KL": divergence,
            **risk_metrics(estimated_risks, true_risks, weights),
            "fitted_k0": parameters[0],
            "fitted_k1": parameters[1],
            "fitted_mean0": parameters[2],
            "fitted_mean1": parameters[3],
            "success": best["success"],
            "projected_gradient": best["projected_gradient"],
            "boundary": at_boundary(best["z"]),
            "within_shape_spread": fit["within_shape_spread"],
            "agreement_fraction": fit["agreement_fraction"],
            "population_zero_KL_check": (
                divergence <= 1e-7
                if method in {"phase_likelihood", "known_shapes"}
                else np.nan
            ),
        })

        for j, history in enumerate(HISTORY_NAMES):
            history_rows.append({
                **context,
                "method": method,
                "history": history,
                "weight": weights[j],
                "true_risk": true_risks[j],
                "estimated_risk": estimated_risks[j],
            })

    range_histories, range_aggregates = explored_risk_ranges(
        correct,
        true_risks,
        weights,
        context,
        population=True,
        true_parameters=true_parameters,
        truth_probabilities=truth_probabilities,
    )

    runs = (
        optimizer_rows(correct, context, False)
        + optimizer_rows(reset, context, True)
    )

    CP.commit(key, {
        "population_accuracy": accuracy_rows,
        "population_histories": history_rows,
        "population_optimizer_runs": runs,
        "population_explored_ranges": range_histories,
        "population_explored_range_summary": range_aggregates,
    })
    CP.save()
    CP.export()

    failed_checks = [
        row["method"]
        for row in accuracy_rows
        if row["method"] in {"phase_likelihood", "known_shapes"}
        and not row["population_zero_KL_check"]
    ]
    if failed_checks:
        print(
            "DIAGNOSTIC WARNING",
            name,
            "population fitting did not reach near-zero KL for",
            failed_checks,
            flush=True,
        )


# ================================================================
# Finite-sample experiment
# ================================================================

def run_sample_task(scenario, n, rep):
    name = scenario["scenario"]
    key = f"sample|{name}|{n}|{rep}"

    if CP.has(key):
        return

    true_parameters = (
        scenario["k0"], scenario["k1"],
        scenario["mean0"], scenario["mean1"],
    )
    true_shapes = true_parameters[:2]

    truth_probabilities = observation_law(true_parameters)
    true_risks, masses = conditional_risks(true_parameters)
    weights = masses / masses.sum()

    observed, _ = simulate(true_parameters, n, reproducible_rng(key))
    counts = np.bincount(
        encode(observed), minlength=len(truth_probabilities)
    )
    frequencies = counts / n

    print("Fitting", key, flush=True)

    correct = fit_family(frequencies, False, key)
    reset = fit_family(frequencies, True, key)
    selected = select_methods(correct, reset, true_shapes)

    context = {
        "scenario": name,
        "k0": scenario["k0"],
        "repair_mean": scenario["mean1"],
        "stage": "sample",
        "n": n,
        "rep": rep,
    }

    # This is a post-fit numerical diagnostic only.
    # The true parameters are not supplied to the estimators.
    truth_objective = objective_value(
        true_parameters, frequencies, reset=False
    )

    accuracy_rows = []
    history_rows = []

    for method, (fit, use_reset) in selected.items():
        best = fit["best"]
        parameters = best["parameters"]

        predicted = observation_law(parameters, reset=use_reset)
        estimated_risks, _ = conditional_risks(
            parameters, reset=use_reset
        )
        gradient_ok = (
            np.isfinite(best["projected_gradient"])
            and best["projected_gradient"] <= PROJECTED_GRADIENT_TOLERANCE
        )

        truth_gap = (
            n * (best["objective"] - truth_objective)
            if method in {"phase_likelihood", "known_shapes"}
            else np.nan
        )

        accuracy_rows.append({
            **context,
            "method": method,
            "record_KL": record_kl(truth_probabilities, predicted),
            **risk_metrics(estimated_risks, true_risks, weights),
            "fitted_k0": parameters[0],
            "fitted_k1": parameters[1],
            "fitted_mean0": parameters[2],
            "fitted_mean1": parameters[3],
            "correct_shapes_selected": bool(
                parameters[:2] == true_shapes
            ),
            "nll": n * best["objective"],
            "success": best["success"],
            "gradient_ok": gradient_ok,
            "projected_gradient": best["projected_gradient"],
            "boundary": at_boundary(best["z"]),
            "within_shape_spread": fit["within_shape_spread"],
            "agreement_fraction": fit["agreement_fraction"],
            "nll_minus_truth_nll": truth_gap,
            "worse_than_true_parameter": (
                truth_gap > 1e-5
                if np.isfinite(truth_gap) else np.nan
            ),
        })

        for j, history in enumerate(HISTORY_NAMES):
            full_history = np.array(
                TARGET_HISTORIES[j] + (0,), dtype=int
            )
            history_rows.append({
                **context,
                "method": method,
                "history": history,
                "true_record_probability": masses[j],
                "weight": weights[j],
                "observed_count": int(np.sum(
                    np.all(observed == full_history, axis=1)
                )),
                "true_risk": true_risks[j],
                "estimated_risk": estimated_risks[j],
                "error": estimated_risks[j] - true_risks[j],
            })

    range_histories, range_aggregates = explored_risk_ranges(
        correct, true_risks, weights, context, population=False
    )

    runs = (
        optimizer_rows(correct, context, False)
        + optimizer_rows(reset, context, True)
    )

    CP.commit(key, {
        "accuracy": accuracy_rows,
        "histories": history_rows,
        "optimizer_runs": runs,
        "explored_ranges": range_histories,
        "explored_range_summary": range_aggregates,
    })

    print("Completed", key, flush=True)


# ================================================================
# Summaries and paired Monte Carlo comparisons
# ================================================================

def summarize_metrics(frame, groups, metrics):
    rows = []

    for key, group in frame.groupby(groups, dropna=False):
        if not isinstance(key, tuple):
            key = (key,)

        for metric in metrics:
            values = pd.to_numeric(group[metric], errors="coerce").dropna().astype(float)

            number = len(values)

            rows.append({
                **dict(zip(groups, key)),
                "metric": metric,
                "mean": float(values.mean()) if number else np.nan,
                "median": float(values.median()) if number else np.nan,
                "q90": float(values.quantile(0.9)) if number else np.nan,
                "mcse": (
                    float(values.std(ddof=1) / np.sqrt(number))
                    if number > 1 else np.nan
                ),
                "replications": number,
            })

    return pd.DataFrame(rows)


def paired_comparisons(frame):
    rows = []
    comparison_methods = [
        "known_shapes", "exponential", "age_reset"
    ]

    for (scenario, n), group in frame.groupby(["scenario", "n"]):
        for metric in ["record_KL", "risk_MSE", "risk_RMSE", "risk_MAE"]:
            wide = group.pivot(index="rep", columns="method", values=metric)

            for comparator in comparison_methods:
                pairs = wide[["phase_likelihood", comparator]].dropna()
                difference = (
                    pairs["phase_likelihood"] - pairs[comparator]
                )
                number = len(difference)

                mcse = (
                    float(difference.std(ddof=1) / np.sqrt(number))
                    if number > 1 else np.nan
                )
                mean = float(difference.mean()) if number else np.nan

                rows.append({
                    "scenario": scenario,
                    "n": n,
                    "metric": metric,
                    "comparison": f"phase_likelihood_minus_{comparator}",
                    "mean_paired_difference": mean,
                    "paired_mcse": mcse,
                    "mc_lower_approx": mean - 1.96 * mcse,
                    "mc_upper_approx": mean + 1.96 * mcse,
                    "paired_replications": number,
                })

    return pd.DataFrame(rows)


def write_summaries():
    frame = CP.frame("accuracy")
    if frame.empty:
        return

    summary = summarize_metrics(
        frame,
        ["scenario", "k0", "repair_mean", "n", "method"],
        [
            "record_KL", "risk_MSE", "risk_RMSE", "risk_bias",
            "risk_MAE", "underreporting_mass", "max_history_error",
            "success", "gradient_ok", "boundary",
            "correct_shapes_selected", "worse_than_true_parameter",
            "agreement_fraction",
        ],
    )
    summary.to_csv(ROOT / "accuracy_summary.csv", index=False)

    paired_comparisons(frame).to_csv(
        ROOT / "paired_comparisons.csv", index=False
    )

    ranges = CP.frame("explored_range_summary")
    if not ranges.empty:
        summarize_metrics(
            ranges,
            ["scenario", "n", "threshold"],
            ["weighted_explored_width", "maximum_explored_width"],
        ).to_csv(ROOT / "explored_width_summary.csv", index=False)

    histories = CP.frame("histories")
    if not histories.empty:
        summarize_metrics(
            histories,
            ["scenario", "n", "method", "history"],
            ["error", "estimated_risk", "observed_count"],
        ).to_csv(ROOT / "history_summary.csv", index=False)

    # Every repetition is retained, including numerically questionable fits.
    # Diagnostics are reported rather than used to silently discard outcomes.
    diagnostics = frame[
        (frame["method"].isin(["phase_likelihood", "known_shapes"]))
        & (
            (~frame["gradient_ok"].astype(bool))
            | frame["boundary"].astype(bool)
            | frame["worse_than_true_parameter"].fillna(False).astype(bool)
        )
    ]
    diagnostics.to_csv(
        ROOT / "fits_requiring_review.csv", index=False
    )


# ================================================================
# Figures
# ================================================================

def plot_metric_by_sample_size(metric, ylabel, filename, logarithmic=False):
    frame = CP.frame("accuracy")
    population = CP.frame("population_accuracy")

    if frame.empty:
        return

    fig, axes = plt.subplots(2, 3, figsize=(14, 8), squeeze=False)

    for index, scenario in enumerate(SCENARIOS):
        ax = axes[index % 2, index // 2]
        name = scenario["scenario"]
        subset = frame[frame.scenario == name]

        for method in METHODS:
            selected = subset[subset.method == method]
            grouped = selected.groupby("n")[metric]
            means = grouped.mean()
            counts = grouped.count()
            errors = grouped.std(ddof=1) / np.sqrt(counts)

            x = means.index.to_numpy(dtype=float)
            y = means.to_numpy(dtype=float)
            e = 1.96 * errors.fillna(0).to_numpy(dtype=float)

            if logarithmic:
                y = np.maximum(y, 1e-12)
                lower = np.minimum(e, y * 0.99)
                yerr = np.vstack([lower, e])
            else:
                yerr = e

            ax.errorbar(
                x, y, yerr=yerr,
                marker="o",
                capsize=3,
                color=COLORS[method],
                label=METHOD_LABELS[method],
            )

            pop = population[
                (population.scenario == name)
                & (population.method == method)
            ]
            if not pop.empty:
                reference = float(pop.iloc[0][metric])
                if not logarithmic or reference > 1e-12:
                    ax.axhline(
                        reference,
                        color=COLORS[method],
                        linestyle=":",
                        alpha=0.65,
                        linewidth=1.2,
                    )

        ax.set_xscale("log")
        if logarithmic:
            ax.set_yscale("log")

        shape_name = {0: "Hyperexponential", 1: "Exponential", 3: "Erlang"}
        ax.set_title(
            f"{shape_name[scenario['k0']]} operation\n"
            f"Mean repair duration {scenario['mean1']}"
        )
        ax.set_xlabel("Number of independent systems")
        ax.set_ylabel(ylabel)
        ax.grid(alpha=0.2)

    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="lower center", ncol=4, frameon=False)
    fig.suptitle(
        "Bars show approximate 95% Monte Carlo uncertainty of means\n"
        "Dotted lines show population-fit diagnostics",
        fontsize=11,
    )
    fig.tight_layout(rect=(0, 0.07, 1, 0.92))
    fig.savefig(ROOT / f"{filename}.png", dpi=300)
    fig.savefig(ROOT / f"{filename}.pdf")
    plt.show()
    plt.close(fig)


def plot_fit_risk_relationship():
    frame = CP.frame("accuracy")
    if frame.empty:
        return

    fig, axes = plt.subplots(2, 3, figsize=(14, 8), squeeze=False)
    largest_n = max(CFG["sample_sizes"])

    for index, scenario in enumerate(SCENARIOS):
        ax = axes[index % 2, index // 2]
        subset = frame[
            (frame.scenario == scenario["scenario"])
            & (frame.n == largest_n)
        ]

        for method in METHODS:
            selected = subset[subset.method == method]
            ax.scatter(
                np.maximum(selected.record_KL, 1e-12),
                selected.risk_RMSE,
                color=COLORS[method],
                label=METHOD_LABELS[method],
                alpha=0.45,
                s=18,
            )

        ax.set_xscale("log")
        ax.set_xlabel("Population record KL divergence")
        ax.set_ylabel("Final-interval risk RMSE")
        ax.set_title(
            f"Operating shape {scenario['k0']}, "
            f"repair mean {scenario['mean1']}"
        )
        ax.grid(alpha=0.2)

    handles, labels = axes[0, 0].get_legend_handles_labels()
    fig.legend(handles, labels, loc="lower center", ncol=4, frameon=False)
    fig.suptitle(f"Observed fit versus hidden risk at n = {largest_n}")
    fig.tight_layout(rect=(0, 0.07, 1, 0.94))
    fig.savefig(ROOT / "fit_vs_risk_by_scenario.png", dpi=300)
    fig.savefig(ROOT / "fit_vs_risk_by_scenario.pdf")
    plt.show()
    plt.close(fig)


def write_readme():
    text = """
EXPERIMENT 1 INTERPRETATION

All systems begin newly in operation.
The target is the probability of at least one failure during the final
inspection interval conditional on each recorded history ending with
two operating readings.

risk_RMSE in accuracy.csv is the square root of a true-probability-weighted
mean squared error across the eligible histories for one simulated dataset.
Its average across repetitions is reported as mean risk_RMSE.
risk_MSE is also retained, so sqrt(mean risk_MSE) can be computed separately.

The known_shapes estimator receives only the true discrete duration shapes.
It estimates both mean durations from the same observed data as other methods.
The other estimators do not receive the true shapes or true means.

Population fits use exact record probabilities and contain no sampling error.
For the correctly specified families, the global population minimum KL is zero.
A failure to approach zero is a numerical diagnostic.
Small nonzero KL with different risk does not prove exact nonidentifiability.

Optimization diagnostics are calculated within each fixed shape pair.
Successful local optimization does not certify a global optimum.
The true-parameter likelihood is evaluated after sample fitting only.
Numerically questionable repetitions are retained and flagged.

Explored risk ranges use evaluated coarse-grid and optimizer candidates.
They are not sharp bounds, confidence intervals, or full likelihood profiles.
Population ranges include the known generating model as a reference.
Sample ranges do not insert the true parameters.

Monte Carlo uncertainty concerns simulation averages and paired comparisons.
It is not uncertainty for the missed-failure risk of a particular observed system.

Local checkpoints do not survive deletion of the Colab runtime.
Download checkpoint.json.gz or the results ZIP to preserve progress.
Resume requires the same configuration and this code version.
If scientific code is edited, update CODE_VERSION before starting a new run.

A completed numerical study does not establish journal suitability by itself.
"""
    (ROOT / "README.txt").write_text(text.strip(), encoding="utf-8")


# ================================================================
# Execute
# ================================================================

metadata = {
    "experiment": "strengthened_experiment_1",
    "code_version": CODE_VERSION,
    "profile": PROFILE,
    "seed": SEED,
    "configuration": CFG,
    "times": TIMES.tolist(),
    "shapes": list(SHAPES),
    "mean_bounds": list(MEAN_BOUNDS),
    "scenarios": SCENARIOS,
    "sample_two_log_likelihood_gaps": list(SAMPLE_TWO_LOG_LIKELIHOOD_GAPS),
    "population_KL_thresholds": list(POPULATION_KL_THRESHOLDS),
    "underreporting_threshold": UNDERREPORTING_THRESHOLD,
}

CP = LocalCheckpoint(ROOT, metadata)

(ROOT / "software_versions.json").write_text(
    json.dumps({
        "python": platform.python_version(),
        "numpy": np.__version__,
        "pandas": pd.__version__,
        "scipy": scipy.__version__,
    }, indent=2),
    encoding="utf-8",
)

write_readme()

try:
    if RUN_VALIDATION:
        run_validation()

    # Population diagnostics are completed before the Monte Carlo study.
    for scenario in SCENARIOS:
        run_population_scenario(scenario)

    # Cycle through scenarios within each repetition so partial runs
    # accumulate evidence across scenarios rather than completing one only.
    for rep in range(CFG["reps"]):
        for n in CFG["sample_sizes"]:
            for scenario in SCENARIOS:
                run_sample_task(scenario, n, rep)

        CP.save()
        CP.export()
        write_summaries()

    plot_metric_by_sample_size(
        "risk_RMSE",
        "Mean final-interval risk RMSE",
        "risk_error_by_sample_size",
    )
    plot_metric_by_sample_size(
        "record_KL",
        "Mean population record KL divergence",
        "record_error_by_sample_size",
        logarithmic=True,
    )
    plot_fit_risk_relationship()

finally:
    CP.save()
    CP.export()
    write_summaries()

print("Completed. Results are in", ROOT)
print("Review population_accuracy.csv and fits_requiring_review.csv first.")
print("Download the checkpoint or ZIP before the runtime is deleted.")

if DOWNLOAD_ON_FINISH:
    download_results()
```
